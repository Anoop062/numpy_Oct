```python
import numpy as np
#create a 4x4 matrix

z=np.array([[1,2,3,4],[3,2,1,5],[9,8,7,6],[4,5,3,2]])

print(z.ndim)

print(z.size)

print(z.shape)

print(z[0])

print(z[1,1])#indexing

print(z[1][1])#same as above

print(z[0:,1],"\n")#print second element from each row

print(z[0:,1:3],"\n")
print(z[2,1:3],"\n")
```

```python
#one matrix
z=np.ones([3,3],dtype=float)
z
```

```python
z=np.zeros([3,3],dtype=float)
z
```

```python
#2X2 in two variables
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a+b,"\n")
print(a-b,"\n")
print(a@b,"\n")
print(a.dot(b))
```

```python
#full matrix
a=np.full([3,3],4)
print(a,"\n")
b=np.full([4,4],9)
b
```

```python
#identity matrix
a=np.identity(9)
print(a)
b=np.eye(4,dtype=int)
print(b,"\n")
```

```python
#
z=np.array([[3.1,5.4,6.7,4.5],[2.4,7.8,6.7,5.9],[2.4,4.5,7.6,9.8]])
print(np.floor(z))
print(np.ceil(z))
print(np.round(z))#nearest even number
```

```python
z=np.array([[3.1,5.4,6.7,4.5],[2.4,7.8,6.7,5.9],[2.4,4.5,7.6,9.8]])
print(z.max())
print(z.min())
print(z.sum())
```

```python
#first max  in each row
z=np.array([[3.1,5.4,6.7,4.5],[2.4,7.8,6.7,5.9],[2.4,4.5,7.6,9.8]])
print(z.max(axis=1))
print(z.max(axis=0))
print(np.sqrt(z))
print(np.cos(z))
print(np.sin(z))
```
